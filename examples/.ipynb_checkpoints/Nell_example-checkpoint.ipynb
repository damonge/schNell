The two main components of `schNell` are `Detector`s and `MapCalculator`s. Let's examine each of them in turn.

## 1. Detectors

`schNell` currently supports 3 types of detectors, all of which inherit from the `Detector` superclass:

- `GroundDetector` objects represent individual 2-arm detectors with fixed positions in Earth coordinates, which therefore rotate and scan with the Earth's motion.
- `GroundDetectorTriangle` objects represent ground-based detectors in triangular configurations (i.e. you will normally spawn three of these together representing the triangle configuration.
- `LISADetector` objects represent LISA-like detectors. I.e. triangular configurations that orbit the Sun on interlacing elliptical orbits.

Let's create some of these

```python
# Two detectors for Hanford and Livignstone.
# These are defined by their Earth coordinates, orientation angle
# and noise PSD (which we read from a file).
hf = snl.GroundDetector('Hanford', 46.4, -119.4, 171.8, '../plots/data/aLIGO.txt')
lv = snl.GroundDetector('Livingstone', 30.7, -90.8, 243.0, '../plots/data/aLIGO.txt')

# A trio of detectors representing the Einstein Telescope
et = [snl.GroundDetectorTriangle(name='ET%d' % i, lat=40.1, lon=9.0,
                                 fname_psd='../plots/data/ET.txt', detector_id=i)
      for i in range(3)]

# A trio of detectors representing the LISA spacecraft
lisa = [snl.LISADetector(i) for i in range(3)]
```

### 1.1 Positions
You can track the positions of the different detectors in their natural coordinates. For instance, let's look at the LISA orbits.

```python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import numpy as np

# First lets get the spacecraft positions throughout the year
t_arr = np.linspace(0, 365*24*3600, 1024)
pos_lisa = np.array([l.get_position(t_arr) for l in lisa])

# We'll plot things in units of 1 A.U.
R_AU = lisa[0].R_AU

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')   # <-- FIXED LINE

for ip, p in enumerate(pos_lisa):
    ax.plot(
        p[0] / R_AU,
        p[1] / R_AU,
        p[2] / R_AU,
        label=r'${\rm %d-th\,\,spacecraft}$' % (ip+1)
    )

ax.set_zlim([-0.1, 0.1])
ax.legend()

plt.show()
```

You can use those positions to check out the effects of breathing on the arm lenghts and angles

```python
# Relative positions
dx01 = pos_lisa[1]-pos_lisa[0]
dx12 = pos_lisa[2]-pos_lisa[1]
dx20 = pos_lisa[0]-pos_lisa[2]

# Lengths
yr = 365*24*3600
plt.figure()
plt.plot(t_arr / yr, np.sqrt(np.sum(dx01**2, axis=0))/2.5E9, label=r'$L_{12}$')
plt.plot(t_arr / yr, np.sqrt(np.sum(dx12**2, axis=0))/2.5E9, label=r'$L_{23}$')
plt.plot(t_arr / yr, np.sqrt(np.sum(dx20**2, axis=0))/2.5E9, label=r'$L_{31}$')
plt.xlabel(r'$t\,[{\rm yrs}]$', fontsize=16)
plt.ylabel(r'$L_{ij} / (2.5\times 10^9 m)$', fontsize=16)
plt.legend();

# Angles
l01 = np.sqrt(np.sum(dx01**2, axis=0))
l20 = np.sqrt(np.sum(dx20**2, axis=0))
theta = np.arccos(np.sum(dx01 * (-dx20), axis=0)/(l01 * l20))
theta *= 3/ np.pi
plt.figure()
plt.plot(t_arr / yr, theta)
plt.xlabel(r'$t\,[{\rm yrs}]$', fontsize=16)
plt.ylabel(r'$\theta_{20} / 60^\circ$', fontsize=16);
```

## 2. NoiseCorrelations
In general, networks of detectors will have correlated noise properties. These are described by `NoiseCorrelator` objects. These are the most useful ones:

 - `NoiseCorrelationConstant` assumes that the correlation matrix $r^{AB}_f=N^{AB}_f/\sqrt{N^{AA}_f N^{BB}_f}$ is constant in frequency. It has two major subclasses: `NoiseCorrelationConstantIdentity` ($r^{AB}=\delta_{AB}$), and `NoiseCorrelationConstantR`, which assumes the same constant correlation coefficient between all pairs of distinct detectors.
 - `NoiseCorrelationFromFunctions` assumes that all pairs of distinct detectors have the same correlation coefficient, which now can be frequency-dependent. Its most important subclass is `NoiseCorrelationLISA`, which implements the specific noise model for the LISA network.

```python
# Let's create two constant correlations
# 1. An identity one
r_eye = snl.NoiseCorrelationConstantIdentity(ndet=2)

# 2. One with a constant correlation coefficient r=0.5
r_0p5 = snl.NoiseCorrelationConstantR(ndet=2, r=0.5)

# As an example, here's how you generate the correlation
# matrix for LISA
r_LISA = snl.NoiseCorrelationLISA(lisa[0])

# Let's then create a pretty plot of the correlation coefficient
# for LISA:
freqs = np.geomspace(1E-4, 1, 1024)
rcorr = r_LISA.get_corrmat(freqs)
plt.figure()
plt.plot(freqs, rcorr[:, 0, 1], 'k-')
plt.xscale('log')
plt.xlabel(r'$f\,[{\rm Hz}]$', fontsize=16)
plt.ylabel(r'$r_f$', fontsize=16);
```

## 3. MapCalculators
Map calculators compute map-level noise quantities for a network of detectors. Let's compute a few basic quantities for the Hanford-Livingstone baseline.

```python
# We will use a pivot frequency of 63 Hz and a spectral index of 2/3,
# corresponding to a stochastic background of binary mergers.
mc = snl.MapCalculator([hf, lv], f_pivot=63., spectral_index=2./3.)

# Let's also create another calculator using the 0.2 correlation
# matrix we created above
mc0p5 = snl.MapCalculator([hf, lv], f_pivot=63.,
                          spectral_index=2./3., corr_matrix=r_0p5)
```

### 3.1 Antenna patterns
Let's start by looking at the antenna patern for this 2-element network. We will sample it on a HEALPix map with resolution parameter `nside=64` (corresponding to degree-sized pixels).

```python
# First, we compute the coordinates for each pixel in the map
nside = 64
npix = hp.nside2npix(nside)
theta, phi = hp.pix2ang(nside, np.arange(npix))

# Let's compute the antenna pattern at the start of the day and in the middle of it.
# We'll do it at a fixed frequency of 40 GHz
# i and j here are the indices of the two detectors we are correlating
antenna_start = np.abs(mc.get_antenna(i=0, j=1, t=0., f=40.,
                                      theta=theta, phi=phi))
antenna_mid = np.abs(mc.get_antenna(i=0, j=1, t=12 * 3500, f=40.,
                                    theta=theta, phi=phi))

# Plot!
hp.mollview(antenna_start)
hp.mollview(antenna_mid)
plt.show()
```

### 3.2 Inverse noise variance maps
You can also compute the inverse-noise variance map for a given timeframe. If you want to e.g. compute the map at the end of an observing season, compute it at a set of timeframes and then add them! The map will be integrated over frequencies with the assumed spectral index.

```python
# We're gonna use a lower nside here, since this calculation can be a bit lengthy
nside = 16
inoi_hl = mc.get_Ninv_t(t=0., f=f_hi, nside=nside, no_autos=True)
# By using no-autos above, we make sure that we don't include any detector auto-correlations

# Plotting time!
hp.mollview(inoi_hl)
```

### 3.3 Noise power spectra
OK, now, the quantity that `schNell` is named after. The noise power spectrum $N_\ell$. We'll compute it assuming one full year of observation.

```python
nside = 64
obs_time = 365 * 24 * 3600
freqs = np.linspace(10., 1010., 101)

nl = mc.get_N_ell(obs_time, freqs, nside, no_autos=True)
# Just for show, let's also calculate the N_ell you'd get
# if both detectors had a 50% correlation coefficient.
nl_0p5 = mc0p5.get_N_ell(obs_time, freqs, nside, no_autos=True)

# Plotting time!
ls = np.arange(3 * nside)
plt.figure()
plt.plot(ls, nl, label=r'$r=0$')
plt.plot(ls, nl_0p5, label=r'$r=0.5$')
plt.xlim([1, 100])
plt.ylim([3E-20, 1E-10])
plt.legend(loc='upper left')
plt.xlabel(r'$\ell$', fontsize=16)
plt.ylabel(r'$N_\ell$', fontsize=16)
plt.loglog();
```

### 3.4 The monopole
Finally, we can also use `schNell` to study the monopole, by computing the so-called power-law-integrated sensitivity curves. Here's what it looks like for the Hanford-Livingstone arm.

```python
nside = 16
pi = mc.get_pi_curve(obs_time, f_hi, nside,
                     no_autos=True, nsigma=2)

# Plotting!
plt.plot(f_hi, pi)
plt.xlabel(r'$f\,[{\rm Hz}]$', fontsize=16)
plt.ylabel(r'$\Omega_{\rm GW}(f)$', fontsize=16)
plt.xlim([10, 1000])
plt.ylim([1E-10, 1E-5])
plt.loglog();
```

```python
import numpy as np
import matplotlib.pyplot as plt
import healpy as hp

from schnell.mapping import MapCalculator
from schnell.detector import LISADetector


# -----------------------------
# Physical constants / settings
# -----------------------------
C_LIGHT = 299_792_458.0  # m/s
L_ARM = 2.5e9            # m (nominal LISA arm length)
F_STAR = C_LIGHT / (2.0 * np.pi * L_ARM)  # transfer frequency f* = c/(2πL)

# H0/h = 100 km/s/Mpc in SI units (s^-1)
MPC_M = 3.0856775814913673e22  # meters
H0_OVER_h = (100.0 * 1000.0) / MPC_M  # s^-1

# Noise model parameters in the paper (Appendix B)
P_OMS_PM = 15.0  # "P" amplitude for OMS in pm
A_ACC_FM = 3.0   # "A" amplitude for acceleration in fm/s^2


# ---------------------------------------------------------
# Appendix B noise functions in "tilde" form (B.14, B.15)
# ---------------------------------------------------------
def S_oms(f_hz):
    """
    Optical Metrology System noise PSD in displacement^2 / Hz.
    Paper parameterization: P in pm, with the (2 mHz / f)^4 rise.
    """
    f = np.asarray(f_hz)
    return (P_OMS_PM * 1e-12) ** 2 * (1.0 + (2.0e-3 / f) ** 4)  # m^2/Hz


def S_acc(f_hz):
    """
    Acceleration noise PSD in (m/s^2)^2 / Hz.
    Paper parameterization: A in fm/s^2, with:
      (1 + (0.4 mHz / f)^2) and (1 + (f / 8 mHz)^4)
    """
    f = np.asarray(f_hz)
    return (A_ACC_FM * 1e-15) ** 2 * (1.0 + (0.4e-3 / f) ** 2) * (1.0 + (f / 8.0e-3) ** 4)


def N_tilde_AE(f_hz):
    """
    \tilde N_{A,E}(f) from Eq. (B.14) (paper).
    Units: 1/Hz (dimensionless PSD).
    """
    f = np.asarray(f_hz)
    x = f / F_STAR
    cosx = np.cos(x)

    # First term: OMS contribution
    term_oms = 0.5 * (2.0 + cosx) * (S_oms(f) / (L_ARM ** 2))

    # Second term: acceleration contribution
    # 2 * [1 + cos(x) + cos^2(x)] * (S_acc / L^2) * (1/(2πf))^4
    term_acc = 2.0 * (1.0 + cosx + cosx ** 2) * (S_acc(f) / (L_ARM ** 2)) * (1.0 / (2.0 * np.pi * f)) ** 4

    return term_oms + term_acc
print(N_tilde_AE(3e-3))


def N_tilde_T(f_hz):
    """
    \tilde N_T(f) from Eq. (B.15) (paper).
    Units: 1/Hz (dimensionless PSD).
    """
    f = np.asarray(f_hz)
    x = f / F_STAR
    cosx = np.cos(x)

    term_oms = (1.0 - cosx) * (S_oms(f) / (L_ARM ** 2))
    term_acc = 2.0 * (1.0 - cosx) ** 2 * (S_acc(f) / (L_ARM ** 2)) * (1.0 / (2.0 * np.pi * f)) ** 4

    return term_oms + term_acc


# -------------------------------------------------------------------
# Helper: Healpy alm index access and full m in [-ell..ell]
# -------------------------------------------------------------------
def alm_full_m(alm, lmax, ell):
    """
    Given healpy alm array (m>=0), return a complex array a_{ell,m}
    for m = -ell..ell using the reality condition for a real map:
      a_{ell,-m} = (-1)^m * conj(a_{ell,m})  for m>0
    """
    out = np.zeros(2 * ell + 1, dtype=np.complex128)
    # m >= 0
    for m in range(0, ell + 1):
        idx = hp.Alm.getidx(lmax, ell, m)
        out[m + ell] = alm[idx]
    # m < 0
    for m in range(1, ell + 1):
        out[ell - m] = ((-1) ** m) * np.conjugate(out[ell + m])
    return out


# -------------------------------------------------------------------
# Compute R~_{lm,11}(f) and R~_{lm,12}(f) from schnell antenna maps
# -------------------------------------------------------------------
def W2(f, tdi="1.5"):
    f = np.asarray(f, dtype=float)
    x = f / F_STAR
    if tdi == "1.0":
        return np.ones_like(f)
    if tdi == "1.5":
        return 4.0 * np.sin(x)**2
    raise ValueError("tdi must be '1.0' or '1.5'")

def map2alm_complex(hp_map, lmax, iter=0):
    """
    Healpy's map2alm is for real maps.
    If the antenna map is complex (because inc_baseline=True),
    compute alms for real and imaginary parts separately:
        alm = alm(Re) + i alm(Im).
    """
    hp_map = np.asarray(hp_map)

    if np.iscomplexobj(hp_map):
        alm_re = hp.map2alm(np.real(hp_map), lmax=lmax, iter=iter)
        alm_im = hp.map2alm(np.imag(hp_map), lmax=lmax, iter=iter)
        return alm_re + 1j * alm_im
    else:
        return hp.map2alm(hp_map, lmax=lmax, iter=iter)

def compute_Rlm_11_12(f_grid, lmax=10, nside=64, t0=0.0, tdi="1.5"):
    """
    Returns:
      alm11_list: list of healpy alm arrays, one per frequency
      alm12_list: list of healpy alm arrays, one per frequency

    These are the *tilde* response coefficients after removing the kernel
    4 (f/f_*)^2 |W(f)|^2 (and after applying the paper's (8π) convention).
    """

    # Build the 3 "detectors" used by schnell's LISA model (the 3 Michelson-like channels)
    dets = [LISADetector(detector_id=i) for i in range(3)]
    mc = MapCalculator(dets)

    # HEALPix sky directions
    npix = hp.nside2npix(nside)
    theta, phi = hp.pix2ang(nside, np.arange(npix))

    f_arr = np.asarray(f_grid, dtype=float)

    # Antenna patterns for (1,1) and (1,2)
    # These are functions on the sky, generally complex if inc_baseline=True.
    ant11 = mc.get_antenna(i=0, j=0, t=t0, f=f_arr, theta=theta, phi=phi,
                           pol=False, inc_baseline=True)
    ant12 = mc.get_antenna(i=0, j=1, t=t0, f=f_arr, theta=theta, phi=phi,
                           pol=False, inc_baseline=True)

    # Some schnell versions return shape [Nf, Npix] and others [Nt, Nf, Npix].
    if ant11.ndim == 3:
        maps11 = ant11[0]  # [Nf, Npix]
        maps12 = ant12[0]
    else:
        maps11 = ant11     # [Nf, Npix]
        maps12 = ant12

    alm11_list = []
    alm12_list = []

    for i_f, f in enumerate(f_arr):

        # --- CRITICAL FIX: complex-safe spherical harmonic transform ---
        alm11 = map2alm_complex(maps11[i_f], lmax=lmax, iter=0)
        alm12 = map2alm_complex(maps12[i_f], lmax=lmax, iter=0)

        # Paper convention: their Eq. (4.15) uses dΩ / (4π),
        # and then they define tilde-R with an 8π in the mapping conventions.
        # This is the same factor you were already using.
        alm11 = alm11 / (8.0 * np.pi)
        alm12 = alm12 / (8.0 * np.pi)

        # Remove the kernel factor to get tilde-R:
        #   kernel = 4 (f/f_*)^2 |W(f)|^2
        kernel = 4.0 * (f / F_STAR) ** 2 * W2(np.array([f]), tdi=tdi)[0]

        alm11 = alm11 / kernel
        alm12 = alm12 / kernel

        alm11_list.append(alm11)
        alm12_list.append(alm12)

    return alm11_list, alm12_list





# -------------------------------------------------------------------
# Build R~_ell^{OO'}(f) from paper’s even/odd formulas (Eq. 4.28 block)
# -------------------------------------------------------------------
import numpy as np
import healpy as hp

def alm_row_mge0(alm, lmax, ell):
    """
    Return a_{ell,m} for m=0..ell from a healpy alm array.
    """
    out = np.zeros(ell + 1, dtype=np.complex128)
    for m in range(0, ell + 1):
        out[m] = alm[hp.Alm.getidx(lmax, ell, m)]
    return out

def sum_over_m_pm_even(weight_mge0, val_mge0):
    """
    Compute sum_{m=-ell}^{ell} weight(m) |val(m)|^2
    when weight(m) is even in m and val(-m) contributes symmetrically.

    Using healpy storage (m>=0):
      total = weight(0)|val(0)|^2 + 2*Σ_{m=1..ell} weight(m)|val(m)|^2
    """
    total = weight_mge0[0] * np.abs(val_mge0[0])**2
    if len(val_mge0) > 1:
        total += 2.0 * np.sum(weight_mge0[1:] * np.abs(val_mge0[1:])**2)
    return total

def Rell_channels_from_alms(R11_alm, R12_alm, lmax, ell):
    """
    Implements Eqs. (4.29)-(4.30) but using only m>=0 healpy alms,
    and doubling m>0 contributions (valid because all weights are even in m).
    """
    m = np.arange(0, ell + 1)  # m>=0 only
    R11 = alm_row_mge0(R11_alm, lmax, ell)
    R12 = alm_row_mge0(R12_alm, lmax, ell)

    cos_2m = np.cos(2.0 * m * np.pi / 3.0)         # even in m
    sin2_m = np.sin(m * np.pi / 3.0) ** 2          # even in m

    out = {}

    if ell % 2 == 1:
        # Eq. (4.29): odd ell
        out["AA"] = 0.0
        out["EE"] = 0.0
        out["TT"] = 0.0

        w_AE = (1.0 + 2.0 * cos_2m) ** 2
        out["AE"] = np.sqrt((1.0 / 3.0) * sum_over_m_pm_even(w_AE, R12))

        w_AT = sin2_m
        out["AT"] = np.sqrt(2.0 * sum_over_m_pm_even(w_AT, R12))
        out["ET"] = out["AT"]

    else:
        # Eq. (4.30): even ell
        phase_m4 = (1.0 + np.exp(-4j * m * np.pi / 3.0))
        phase_p2 = (1.0 + np.exp( 2j * m * np.pi / 3.0))

        val_AA = phase_m4 * R11 - 2.0 * R12
        out["AA"] = np.sqrt((1.0 / 4.0) * sum_over_m_pm_even(np.ones_like(m), val_AA))
        out["EE"] = out["AA"]

        w_TT = (1.0 + 2.0 * cos_2m) ** 2
        val_TT = (R11 + 2.0 * R12)
        out["TT"] = np.sqrt((1.0 / 9.0) * sum_over_m_pm_even(w_TT, val_TT))

        w_AE = sin2_m
        val_AE = phase_p2 * R11 - 2.0 * R12
        out["AE"] = np.sqrt((1.0 / 3.0) * sum_over_m_pm_even(w_AE, val_AE))

        w_AT = sin2_m
        val_AT = phase_p2 * R11 + R12
        out["AT"] = np.sqrt((2.0 / 3.0) * sum_over_m_pm_even(w_AT, val_AT))
        out["ET"] = out["AT"]

    return out


# -------------------------------------------------------------------
# Eq. (4.42) and (4.43): channel-channel and combined sensitivity
# -------------------------------------------------------------------
def omega_channel_channel(f, Rell_OO, Ntilde_O, Ntilde_O2, f3_in="numerator"):
    """
    Toggle where f^3 goes. This is the critical scaling that determines whether Ω has a bathtub.
    f3_in: "numerator" or "denominator"
    """
    pref = (4.0 * np.pi**2 * np.sqrt(4.0 * np.pi)) / (3.0 * (H0_OVER_h ** 2))
    base = np.sqrt(Ntilde_O * Ntilde_O2) / Rell_OO

    if f3_in == "numerator":
        return pref * (f ** 3) * base
    elif f3_in == "denominator":
        return pref * base / (f ** 3)
    else:
        raise ValueError("f3_in must be 'numerator' or 'denominator'")




def omega_combined_from_channels(omega_list):
    """
    Eq. (4.43) optimal combination:
      Ω^ell_{GW,n}(f) h^2 = [ Σ_{OO'} (1 / Ω^2_{GW,OO',n}) ]^{-1/2}
    """
    inv2 = np.zeros_like(omega_list[0])
    for om in omega_list:
        inv2 += 1.0 / (om ** 2)
    return inv2 ** (-0.5)


# -----------------------------
# Main’: reproduce Figure 9
# -----------------------------
def recreate_figure9(f_grid, lmax=10, nside=64):
    """
    Recreate Fig. 9 curves using:
      - response multipoles from schnell -> Rell_channels_from_alms (Eqs. 4.29–4.30 block)
      - tilde-noise curves from Appendix B: N_tilde_AE, N_tilde_T
      - optimal combination across (OO') per Eq. (4.43)

    IMPORTANT FIX:
      With tilde-noise, the Fig. 9 bathtub requires Omega_n ∝ sqrt(Ntilde)/[f^3 * Rtilde].
      (This is the only change relative to your current function.)
    """

    f_grid = np.asarray(f_grid, dtype=float)

    # 1) Compute multipole coefficients (healpy alm arrays) for the two independent baselines
    alm11_list, alm12_list = compute_Rlm_11_12(
        f_grid, lmax=lmax, nside=nside, t0=0.0, tdi="1.5"
    )

    # 2) Paper tilde-noise functions (Appendix B)
    N_A = N_tilde_AE(f_grid)
    N_E = N_A.copy()
    N_T = N_tilde_T(f_grid)
    N_map = {"A": N_A, "E": N_E, "T": N_T}

    # 3) Y00 rescaling used in the figure
    Y00 = 1.0 / np.sqrt(4.0 * np.pi)

    # --- local “correct for tilde-noise” omega definition ---
    def omega_channel_channel_tilde(f, Rtilde_ell, Ntilde1, Ntilde2):
        """
        Use the form consistent with tilde-noise:
          Ω_n h^2 = pref * sqrt(Ntilde1*Ntilde2) / ( f^3 * Rtilde_ell )
        This produces the low-f wall (bathtub) because Ntilde ~ f^-4 at low f.
        """
        pref = (4.0 * np.pi**2 * np.sqrt(4.0 * np.pi)) / (3.0 * (H0_OVER_h ** 2))
        return pref * (np.sqrt(Ntilde1 * Ntilde2) / ((f ** 3) * Rtilde_ell))

    omega_ell = {}

    for ell in range(lmax + 1):

        # 4) Build R~^ell_{OO'}(f) using your even/odd formulas
        Rell = {key: np.zeros_like(f_grid, dtype=float) for key in ["AA", "EE", "TT", "AE", "AT", "ET"]}

        for i_f in range(len(f_grid)):
            Rch = Rell_channels_from_alms(
                alm11_list[i_f],
                alm12_list[i_f],
                lmax,
                ell
            )
            for k, v in Rch.items():
                Rell[k][i_f] = float(np.real(v))

        # 5) Build channel-channel Ω_n curves and combine per Eq. (4.43)
        omega_terms = []

        # Auto channels: AA, EE, TT
        for O in ["A", "E", "T"]:
            key = O + O
            Rv = Rell[key]
            om = np.full_like(f_grid, np.inf, dtype=float)
            mask = Rv > 0
            om[mask] = omega_channel_channel_tilde(
                f_grid[mask], Rv[mask], N_map[O][mask], N_map[O][mask]
            )
            omega_terms.append(om)

        # Cross channels: AE, AT, ET
        for (O1, O2, key) in [("A", "E", "AE"), ("A", "T", "AT"), ("E", "T", "ET")]:
            Rv = Rell[key]
            om = np.full_like(f_grid, np.inf, dtype=float)
            mask = Rv > 0
            om[mask] = omega_channel_channel_tilde(
                f_grid[mask], Rv[mask], N_map[O1][mask], N_map[O2][mask]
            )
            omega_terms.append(om)

        omega_comb = omega_combined_from_channels(omega_terms)

        # 6) Apply Y00 plotting rescale
        omega_ell[ell] = omega_comb * Y00

    return omega_ell


if __name__ == "__main__":
    # Frequency range similar to Figure 9 (log-spaced)
    f = np.logspace(-4, -0.3, 160)

    curves = recreate_figure9(f, lmax=10, nside=64)

    plt.figure(figsize=(8, 6))
    for ell in range(0, 11):
        plt.loglog(f, curves[ell], label=fr"$\ell={ell}$")

    plt.xlabel("f [Hz]")
    plt.ylabel(r"$\Omega^{\ell}_{\rm GW,n}(f)\,h^2 \times Y_{00}$")
    plt.title("Recreation of Figure 9: LISA multipole sensitivity")
    plt.grid(True, which="both", ls=":")
    plt.legend(ncol=2, fontsize=9)
    plt.tight_layout()
    plt.show()
```
